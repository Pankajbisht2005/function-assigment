Q1  What is the difference between a function and a method in Python?
- Function

 A function is a block of reusable code that performs a specific task.

It is defined using def keyword (or sometimes lambda).

It is not tied to any object/class, and can be called independently

```
# Example:
def greet(name):
    return f"Hello, {name}!"

print(greet("Pankaj"))
```
- Method:

A method is also a function, but the difference is that it belongs to an object.

Defined inside a class.

Always takes self as the first parameter.

Methods are invoked on objects.



```
# Example:
class Person:
    def __init__(self, name):
        self.name = name

    # Method (because it's inside a class)
    def greet(self):
        return f"Hello, {self.name}!"

p = Person("Pankaj")

print(p.greet())

```

Q5 What are iterators in Python and how do they differ from iterables?
- iterators:

 An iterator is an object that actually performs the iteration.

It keeps track of the current position.

It must implement two methods:

__iter__()  returns the iterator object itself

__next__()  returns the next element, and raises StopIteration when no items are left.


```
#Example

it = iter(my_list)

print(next(it))  
print(next(it))  
print(next(it))  

```
- iterables:

An iterable is any Python object that can be looped over (iterated).

It has an __iter__() method that returns an iterator.

Examples: lists, tuples, strings, sets, dictionaries, ranges.



```

my_list = [1, 2, 3]

for item in my_list:
    print(item)

```

Q7 What are the advantages of using generators over regular functions?
- Regular function:

 A regular function in Python runs its code and returns a single value using return. Once returned, the function’s state is lost.


 -generators:

It uses the yield keyword instead of return. It produces a sequence of values one at a time and remembers its state between calls, allowing iteration without storing everything in memory.


```
# Example of generators

def generate_squares(n):
    for i in range(n):
        yield i ** 2

gen = generate_squares(5)
for value in gen:
    print(value)

```

Q10  What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
- 1. map() Function


Applies a given function to each element of an iterable.

Returns a map object (iterator) with the results.

Does not filter elements; it transforms them.


```
#Example:
 nums = [1, 2, 3, 4]
squared = map(lambda x: x ** 2, nums)
print(list(squared))  
output: [1, 4, 9, 16]

```
- 2. filter() Function

Filters elements of an iterable based on a condition function that returns True/False.

Returns a filter object (iterator) with elements that satisfy the condition.


```
#Example
nums = [10, 15, 20, 25]
evens = filter(lambda x: x % 2 == 0, nums)
print(list(evens))  
Output: [10, 20]

```
3. reduce() Function


reduce() is from functools module.

Applies a function cumulatively to elements of an iterable to reduce it to a single value.


```
# Example:
from functools import reduce
nums = [1, 2, 3, 4]
sum_all = reduce(lambda x, y: x + y, nums)
print(sum_all)
Output: 10

```

````python
 #1 Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.
 def sum_even_numbers(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:   # Check if the number is even
            total += num    # Add it to total
    return total
nums = [10, 15, 20, 25, 30]
print("Sum of even numbers:", sum_even_numbers(nums))

````

````python
#2 Create a Python function that accepts a string and returns the reverse of that string.
def reverse_string(s):
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str

text = "Pankaj"
print("Reversed string:", reverse_string(text))
````

````python
#4 Write a Python function that checks if a given number is prime or not from 1 to 200
# Function to check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False  # Divisible by a number  Not prime
    return True  # Prime if not divisible by any number

# Check primes from 1 to 200
print("Prime numbers from 1 to 200:")
for num in range(1, 201):
    if is_prime(num):
        print(num, end=" ")

````

````python
#5  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
# Generate first 10 Fibonacci numbers
class Fibonacci:
    def __init__(self, n_terms):
        self.n_terms = n_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return self.a
        elif self.count == 1:
            self.count += 1
            return self.b
        else:
            next_value = self.a + self.b
            self.a, self.b = self.b, next_value
            self.count += 1
            return next_value

fib_sequence = Fibonacci(10)
for num in fib_sequence:
    print(num, end=" ")

````

````python
#6 Write a generator function in Python that yields the powers of 2 up to a given exponent
# Generator function to yield powers of 2
def powers_of_two(max_exponent):

    for i in range(max_exponent + 1):
        yield 2 ** i  # Yield 2^i

# Input from user
n = int(input("Enter the maximum exponent: "))

# Using the generator to print powers of 2
print(f"Powers of 2 from 0 to {n}:")
for value in powers_of_two(n):
    print(value, end=" ")


````

````python
#7 Implement a generator function that reads a file line by line and yields each line as a string.
# 1. Create a sample file
with open("example.txt", "w") as f:
    f.write("Hello Dost\n")
    f.write("This is a test file.\n")
    f.write("Have a nice day!\n")

# 2. Generator function
def read_file_lines(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()

# 3. Read the file using generator
for line in read_file_lines("example.txt"):
    print(line)


````
